```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("students.csv")

# Display the complete dataset
print("Dataset:")
print(df)
```

```python
df.columns
```

```python
#indexing, we can perform indexing to access specific rows and columns in the DataFrame. For example, we can access the first row and the 'Payment_Method' column using the following code: 
df.iloc[0]['Payment_Method']
```

```python
#indexing, we can also access specific columns using their names. For example, we can access the 'order_id' and 'Customer_ID' columns using the following code:
df[['Order_ID','Customer_ID']]
```

```python
#We can also use the `isin()` method to filter the DataFrame based on multiple values in a column. For example, we can filter the DataFrame to include only rows where the 'Payment_Method' is either 'PayPak' or 'Credit Card' using the following code:
df[df['Payment_Method'] == 'PayPak']
```

```python
#We can also use the `isin()` method to filter the DataFrame based on multiple values in a column. For example, we can filter the DataFrame to include only rows where the 'Payment_Method' is either 'PayPak' or 'Credit Card' using the following code:   
df[df['Payment_Method'].isin(['PayPak','Credit Card'])]
```

```python
#We can also use the `~` operator to filter the DataFrame to exclude rows where the 'Payment_Method' is either 'PayPak' or 'Credit Card' using the following code:
df[~df['Payment_Method'].isin(['PayPak','Credit Card'])]
```

```python
#We can also use the `loc` method to access specific rows and columns in the DataFrame based on a condition. For example, we can access the row where the 'Customer_ID' is 'CUST1050' using the following code:
df.loc[df['Customer_ID']=='CUST1050', 'Category'] = 'Office Supplies'
```

```python
df.loc[df['Customer_ID']=='CUST1050']
```

```python
#We can use the `dropna()` method to remove rows with missing values from the DataFrame:
df.dropna()
```

```python
#We can use the `fillna()` method to fill missing values in the DataFrame with a specified value. For example, we can fill missing values in the 'Order_ID' column with 0 using the following code:
df.fillna({'Order_ID': 0}, inplace=True)
```
