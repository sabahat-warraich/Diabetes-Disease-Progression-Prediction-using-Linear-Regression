```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("insurance.csv")
df.head()
```

```python
plt.scatter(df.Age, df.Purchased, marker='+', color='red')
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['Age']], df['Purchased'], test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
X_test
```

```python
model.predict(X_test)
```

```python
y_pred_proba =model.predict_proba(X_test)
print(y_pred_proba)
```

```python
# -----------------------------
# Visualize Logistic Regression
# -----------------------------

# Create age values for the curve
age_range = np.linspace(
    df['Age'].min(),
    df['Age'].max(),
    100
).reshape(-1, 1)

# Predict probability for each age
probabilities = model.predict_proba(age_range)[:, 1]

# Plot actual data
plt.scatter(
    X_test['Age'],
    y_test,
    marker='o',
    label='Actual'
)

# Plot predicted probability
plt.plot(
    age_range,
    probabilities,
    label='Predicted Probability'
)

plt.xlabel("Age")
plt.ylabel("Probability of Purchase")
plt.title("Logistic Regression: Age vs Purchase Probability")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# -----------------------------
# Logistic Regression Loss
# -----------------------------
#binary log-loss formula:
# −1/n∑[ylog(p)+(1−y)log(1−p)]

def log_loss(y_true, y_pred):

    n = len(y_true)

    loss = -(1/n) * np.sum(
        y_true * np.log(y_pred)
        + (1 - y_true) * np.log(1 - y_pred)
    )

    return loss


loss = log_loss(
    y_test.values,
    y_pred_proba[:, 1]
)

print("Log Loss:", loss)
```
