```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("santander.csv", nrows=10000)
```

```python
df.shape
```

```python
df.head()
```

```python
X=df.drop('TARGET', axis=1)
Y=df['TARGET']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
X_train.shape, X_test.shape
```

```python
# Try different thresholds to see the effect
for threshold in [0, 0.01, 0.05, 0.1]:
    selector = VarianceThreshold(threshold=threshold)
    X_selected = selector.fit_transform(X_train)
    
    # How many features dropped?
    dropped = X_train.shape[1] - X_selected.shape[1]
    print(f"Threshold: {threshold} → Features kept: {X_selected.shape[1]} (Dropped: {dropped})")
```

```python
selector = VarianceThreshold(threshold=0)
X_selected = selector.fit_transform(X_train)
```
